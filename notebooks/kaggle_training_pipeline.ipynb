## 4. Run Training with Headless Commit Support
This code executes multi-epoch model training using Negative Pearson Loss and Cosine Spectral Distance, saving PyTorch checkpoints and ONNX models into `/kaggle/working`.

```python
import sys
sys.path.insert(0, '/kaggle/working/rppg-vital-sign-monitor/backend')

from app.models_dl.trainer import PhysNetTrainer

trainer = PhysNetTrainer(
    device="cuda" if torch.cuda.is_available() else "cpu",
    checkpoint_dir=KAGGLE_CHECKPOINTS,
)

summary = trainer.run_training_experiment(
    num_epochs=15,
    num_subjects=30,
    batch_size=4,
    experiment_name="Kaggle_Dual_T4_PhysNet_Benchmark",
    backup_dir=KAGGLE_CHECKPOINTS,
)

print("Kaggle Training Experiment Complete!")
print(f"Validation Loss: {summary['final_val_loss']}")
print(f"Test Loss: {summary['test_loss']}")
print(f"Best Checkpoint: {summary['best_checkpoint_path']}")
print(f"ONNX Model: {summary['onnx_export_path']}")
```
