# Cloud-Based 1-Click Android APK Builder for rPPG Vital Signs Monitor

**Project:** Development and Validation of an rPPG-Based System for Non-Contact Vital Sign Monitoring  
**Cloud Infrastructure:** Google Colab (100% Free CPU/GPU Runtime) + Google Drive  
**Objective:** Build and export the standalone Android APK (`app-release.apk`) in the cloud without requiring Flutter or Android SDK installation on your personal computer.

---

### Step 1: Mount Google Drive
Mount your Google Drive to automatically save the compiled APK into `RPPG_FYP/mobile_apk/`.

```python
from google.colab import drive
import os

drive.mount('/content/drive')
drive_output_dir = '/content/drive/MyDrive/RPPG_FYP/mobile_apk'
os.makedirs(drive_output_dir, exist_ok=True)
print(f"Target Google Drive APK directory: {drive_output_dir}")
```

### Step 2: Install Flutter SDK in Google Colab
Downloads and extracts the stable Flutter Linux SDK into the Colab environment (~1 minute).

```python
%%bash
# Download Flutter SDK stable channel
if [ ! -d "/content/flutter" ]; then
    echo "Downloading Flutter SDK..."
    curl -s -O https://storage.googleapis.com/flutter_infra_release/releases/stable/linux/flutter_linux_3.24.5-stable.tar.xz
    echo "Extracting Flutter SDK..."
    tar -xf flutter_linux_3.24.5-stable.tar.xz -C /content/
    rm flutter_linux_3.24.5-stable.tar.xz
fi

export PATH="/content/flutter/bin:$PATH"
flutter doctor -v
```

### Step 3: Clone or Sync the Project Repository
If your repository is on GitHub, enter your repository URL below. Otherwise, you can upload your `mobile/` directory directly into Colab.

```python
import os

# Set your repository URL if hosted on GitHub
REPO_URL = "https://github.com/YOUR_USERNAME/YOUR_REPOSITORY.git"  # Update with your repo if applicable

if os.path.exists("/content/project"):
    print("Project directory already exists at /content/project")
else:
    if "YOUR_USERNAME" not in REPO_URL:
        !git clone $REPO_URL /content/project
    else:
        print("Creating local project directory structure...")
        os.makedirs("/content/project", exist_ok=True)
        print("Please upload your 'mobile' folder into /content/project/ or provide your GitHub repo URL.")
```

### Step 4: Build Release Android APK (`flutter build apk --release`)
Compiles the production APK with Android camera and networking permissions enabled.

```python
%%bash
export PATH="/content/flutter/bin:$PATH"
cd /content/project/mobile

echo "Fetching Flutter dependencies..."
flutter pub get

echo "Compiling Android Release APK..."
flutter build apk --release

ls -lh build/app/outputs/flutter-apk/app-release.apk
```

### Step 5: Save APK to Google Drive & Trigger Browser Download
Copies the generated `app-release.apk` directly to your Google Drive and downloads it straight to your computer.

```python
import shutil
from google.colab import files

apk_source = "/content/project/mobile/build/app/outputs/flutter-apk/app-release.apk"
apk_dest_drive = "/content/drive/MyDrive/RPPG_FYP/mobile_apk/rppg_vital_signs_release.apk"

if os.path.exists(apk_source):
    # 1. Copy to Google Drive permanent storage
    shutil.copyfile(apk_source, apk_dest_drive)
    print(f"Successfully saved APK to Google Drive: {apk_dest_drive}")
    
    # 2. Trigger browser download
    print("Triggering browser download...")
    files.download(apk_source)
else:
    print(f"Error: APK file not found at {apk_source}. Please verify build logs above.")
```
